```python
weight = float(input("Enter your weight in Kg: "))
height = float(input("Enter your height in cms: "))

BMI = weight / ((height / 100) ** 2)

if BMI < 18.5:
    category = "Underweight"
elif BMI < 25:
    category = "Normal weight"
elif BMI < 30:
    category = "Overweight"
else:
    category = "Obese"

print("Your BMI is:", BMI)
print("Category:", category)
```
